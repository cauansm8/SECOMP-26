### Processamento Digital de Imagens (relembrando)

As imagens são matrizes de pixels. Cada pixel armazena um valor numérico que representa a intensidade da iluminação naquele ponto da imagem.

Em imagens coloridas, elas são representadas por 3 matrizes sobrepostas (cada matriz representa a intensidade de um dos componentes de iluminação RGB)

É possível fazer operações aritméticas e lógicas (Algebra Linear), inclusive as operações convolucionais.

### Operações Convolucionais

Definição: uma camada convolucional é o bloco central de uma CNN. Cada convolução consiste numa operação matemática que sobrepõe suas funções medindo o cálculo de integral da sua multiplicação pontual.

### Arquitetura Genérica de uma CNN

- Entrada;
- Convolução;
- Pooling;
- Convolução;
- Pooling;
- Totalmente conectada.

### Rectified Linear Unit (ReLU)

Função de ativação amplamente utilizada ao projetar CNNs. Ela introduz características não lineares (mais detalhes).

$$ f(x) = max (0, x) $$

Introduz exponenciação, ou seja, a cada camada a obtenção de características mais complexas melhora.

Usamos mais esta.

### Leaky ReLU

Versão modificada da ReLU. 

Permite um valor pequeno e diferente de zero quando a entrada é negativa.

### Camada Pooling

Reduz as dimensões dos features maps, mantendo apenas as informações mais relevantes (depende da função de pooling)

Exemplo de pooling: max pooling, average pooling.

(90% do trabalho com CNN é pré-processamento)

### Flatten e Fully-Conected Layers

Os feature maps (todas as características da imagem ficam nesse array) resultantes passam por um Flatten, que os transforma em um vetor unidimensional que é então alimentada a uma rede neural Fully Connected Layer, responsável por aprender padrões complexos a partir das features extraídas e, finalmente, classificar a imagem.

### Vanishing Gradient

Em redes muito profundas, o gradiente que é propagado de volta durante o treinamento tende a se reduzir progressivamente à medida que atravessa muitas camadas.

## Arquiteturas CNN

### ResNet

Aborda o problema da degradação e de desaparecimento do gradiente em redes muito profundas, através do uso de blocos residuais, que permitem a construção de conexões diretas entre camadas.

### Inception V3
No processo de convolução, ao invés de escolher um único kernel, podemos usar vários kernels (sobrepondo os kernels). Muito usado em problemas com vários tipos de classificação.

## Prática 1

https://github.com/BrunoAFurquim/Secompp-CNN

anotações gerais:
- data augmentation: gera novas imagens a partir das imagens do dataset (rotação, zoom, ...)
- conjuntos de treino e teste
- conjunto homogêneo: todas as classes têm quantidades semelhantes de imagens
- overfitting: treino excessivo do modelo (deixa de ser genérico - decora o dataset). O conjunto de validação ajuda a descobrir o overfitting.
- underfitting: treino insuficiente do modelo (não consegue classificar)
- precisa saber escolher o número de épocas
- pré-processamento: normalizar os valores (0 - 1), ajuda no processamento da rede neural
- one hot-encoding: para as classes
- define as convoluções, tamanho do kernel, pooling, dropout (desliga % dos neurônios da camada para evitar overfitting)
- learning_rate e decay: define a velocidade do aprendizado
- checkpoints: servem para evitar a perda total do aprendizado (caso caia a energia)
- é importante comparar as acurácias de treino e validação de acordo com as épocas para encontrar se está ocorrendo overfitting (cresce a acurácia de teste e a acurácia de validação estagna ou diminui)
- a matriz de confusão mostra quais classes o modelo acertou e errou (exemplo do avião e pássaro)
